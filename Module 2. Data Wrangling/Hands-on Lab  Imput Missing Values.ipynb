### Step 1: Import Required Libraries


```python
import pandas as pd
```

### Step 2: Load the Dataset Into a Dataframe


#### **Read Data**
<p>

```python
file_path = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
df.head()
```

### Step 3. Finding and Removing Duplicates
##### Task 1: Identify duplicate rows in the dataset.


```python
## Write your code here
df.head(3)
```

```python
df.duplicated().sum()
```

```python
df[df.duplicated()]
```

##### Task 2: Remove the duplicate rows from the dataframe.



```python
print("Before: {}".format(df.shape))

df.drop_duplicates(inplace=True)

print("After: {}".format(df.shape))
print("Duplicates left: {}".format(df.duplicated().sum()))
```

### Step 4: Finding Missing Values
##### Task 3: Find the missing values for all columns.


```python
## Write your code here
df.isna().head(3)
```

```python
df.isnull().sum()
```

```python
import pandas as pd

# 1. Missing values count per column
df.isnull().sum()

# 2. Only columns that have missing values
df.isnull().sum()[df.isnull().sum() > 0]

# 3. With percentage
missing = pd.DataFrame({
    'Missing Count': df.isnull().sum(),
    'Percentage': (df.isnull().sum() / len(df) * 100)
})
missing[missing['Missing Count'] > 0].sort_values(by='Missing Count', ascending=False)
```

##### Task 4: Find out how many rows are missing in the column RemoteWork.


```python
## Write your code here
df["RemoteWork"].isna().sum()
print("{} are missing".format(df["RemoteWork"].isna().sum()))
```

### Step 5. Imputing Missing Values
##### Task 5: Find the value counts for the column RemoteWork.


```python
## Write your code here
df["RemoteWork"].value_counts()
```

##### Task 6: Identify the most frequent (majority) value in the RemoteWork column.



```python
## Write your code here
df["RemoteWork"].mode()[0]
```

##### Task 7: Impute (replace) all the empty rows in the column RemoteWork with the majority value.



```python
## Write your code here
df["RemoteWork"]=df["RemoteWork"].fillna(df["RemoteWork"].mode()[0])
```

```python
df["RemoteWork"].value_counts()
```

##### Task 8: Check for any compensation-related columns and describe their distribution.



```python
import pandas as pd


# Check if 'comp' is in any column name (lowercase)
comp_cols = [col for col in df.columns if 'comp' in col.lower()]

print("Found: {}".format(comp_cols))

# Describe those columns
for col in comp_cols:
    print("\n--- {} ---".format(col))
    print(df[col].describe())
```

```python
df["CompTotal"].isna().sum()
print("{} are missing".format(df["CompTotal"].isna().sum()))
print("\n\n")
print(df["CompTotal"].value_counts())
## Write your code here
df["CompTotal"].mode()[0]
## Write your code here
df["CompTotal"]=df["CompTotal"].fillna(df["CompTotal"].mode()[0])
print("Missing values are replaced by:  {}".format(df["CompTotal"].mode()[0]))
print("\n\nNew Distribution:\n\n {}".format(df["CompTotal"].value_counts()))
```
