```python
import pandas as pd

df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv")
df.drop_duplicates(inplace=True)

# 1. Check missing values
missing_percent = df['ConvertedCompYearly'].isnull().mean()*100
print("The missing values in ConvertedCompYearly is {} percent".format(missing_percent))

# 2. Handle them
df_clean = df.dropna(subset=['ConvertedCompYearly'])

# Now analysis on clean data
print("Median compensation is {}".format(df_clean['ConvertedCompYearly'].median()))
print("Mean compensation is {}".format(df_clean['ConvertedCompYearly'].mean()))
```
