### Step 3: Handle Missing Data


- Identify missing values in the dataset.

- Impute or remove missing values as necessary to ensure data completeness.



```python
## Write your code here
# missing count per column
missing = df.isnull().sum()
print(missing[missing > 0].sort_values(ascending=False))

# percentage
print("\n--- % missing ---")
print((df.isnull().sum() / len(df) * 100).sort_values(ascending=False).round(2))

# total missing
print(f"\nTotal missing cells: {df.isnull().sum().sum()}")

# heatmap if you have seaborn
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(12,6))
sns.heatmap(df.isnull().sample(10), cbar=False)
plt.show()
```

```python
# For your survey data, best strategy:
print(f"Initial missing: {df.isnull().sum().sum()}")
# 1. Drop rows where Main target columns are missing
df_clean = df.dropna(subset=['Employment', 'EdLevel'])

# 2. For categorical columns with few missing, fill with 'Not Answered' / Mode
for col in ['Employment', 'EdLevel', 'RemoteWork', 'Country']:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].mode()[0])

# 3. For numerical columns (like YearsCode, Salary), fill with median
for col in df.select_dtypes(include='number').columns:
    df[col] = df[col].fillna(df[col].median())

# 4. If a column has > 60% missing - just drop the column
thresh = len(df) * 0.4 # keep columns with at least 40% data
df = df.dropna(thresh=thresh, axis=1)

df_clean = df.copy()

print(f"Before: {len(df)} -> After cleaning: {len(df_clean)}")
print(f"Remaining missing: {df_clean.isnull().sum().sum()}")
```

```python
#KDE , transform jobsat to numeric
df['JobSat'] = pd.to_numeric(df['JobSat'], errors='coerce')

# then KDE directly
import seaborn as sns
sns.kdeplot(df['JobSat'].dropna(), fill=True)
```

```python
# For a 1-10 JobSat KDE, here's how to interpret it:

# 1. Central Tendency:
# The curve peaks around 7-8. That means the most common satisfaction level is 7-8/10. Average developer is quite satisfied.

# 2. Skew:
# The plot is left-skewed (negatively skewed). The long tail is on the left (1-4). Few people are very dissatisfied, but that tail pulls the mean down slightly below the peak.

# 3. Spread:
# Most values are clustered between 6-9. Very tight distribution. Means most devs agree on being satisfied. Scores 1-3 are rare outliers.

# 4. Trend:
# Job satisfaction is high overall but not perfect. People aren't giving 10s. They give 7-8, which means satisfied but with room for improvement (work-life balance, salary, tech stack).
```

### Step 6: Programming Languages Analysis


- Compare the frequency of programming languages in `LanguageHaveWorkedWith` and `LanguageWantToWorkWith`.
  
- Visualize the overlap or differences using a Venn diagram or a grouped bar chart.


```python
## Write your code 
df[["LanguageHaveWorkedWith","LanguageWantToWorkWith"]].head(10)
```

```python
import pandas as pd

# Load
url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(url)

# Count languages - Have
have = df['LanguageHaveWorkedWith'].dropna().str.split(';').explode().str.strip().value_counts()

# Count languages - Want
want = df['LanguageWantToWorkWith'].dropna().str.split(';').explode().str.strip().value_counts()

# Compare side by side
comparison = pd.DataFrame({'Have': have, 'Want': want}).fillna(0).astype(int)
comparison['Gap'] = comparison['Want'] - comparison['Have']

# Show results
print(comparison.sort_values('Have', ascending=False).head(15))
```

```python

comparison.head(10).plot(kind='bar', figsize=(12,6))
plt.title('Top 10: Have vs Want')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```

### Step 8: Correlation between Job Satisfaction and Experience

### Step 9: Cross-tabulation Analysis (Employment vs. Education Level)


- Analyze the relationship between employment status (`Employment`) and education level (`EdLevel`).

- **Instruction**: Create a cross-tabulation using `pd.crosstab()` and visualize it with a stacked bar plot if possible.


```python
df[["EdLevel"]].value_counts()
```

```python
## Write your code here
#using heatmap
# using heatmap
import seaborn as sns
import matplotlib.pyplot as plt

top_emp = df['Employment'].value_counts().head(5).index
filtered = df[df['Employment'].isin(top_emp)]

table = pd.crosstab(filtered['Employment'], filtered['EdLevel'], normalize='index')

plt.figure(figsize=(12,7))
sns.heatmap(table, annot=True, fmt=".2f", cmap="Blues")

plt.title("Top 5 Employment vs All EdLevel")
plt.xticks(rotation=45, ha='right')  # <-- fix here
plt.yticks(rotation=0)
plt.tight_layout()  # <-- and here, so labels don't cut
plt.show()
```

```python
df["EdLevel"].value_counts()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

top_emp = df['Employment'].value_counts().head(5).index
filtered = df[df['Employment'].isin(top_emp)]

# 1. DEFINE the logical order for EdLevel - change these names to match YOUR data exactly
ed_order = [
    "Primary/elementary school",
    "Secondary school (e.g. American high school, German Realschule or Gymnasium, etc.)",
    "Some college/university study without earning a degree",
    "Associate degree (A.A., A.S., etc.)",
    "Bachelor’s degree (B.A., B.S., B.Eng., etc.)",
    "Master’s degree (M.A., M.S., M.Eng., MBA, etc.)",
    "Professional degree (JD, MD, Ph.D, Ed.D, etc.)",
    "Something else"
]

# keep only orders that actually exist in your data
ed_order = [x for x in ed_order if x in filtered['EdLevel'].unique()]

# 2. make it ordered category
filtered['EdLevel'] = pd.Categorical(filtered['EdLevel'], categories=ed_order, ordered=True)

# 3. crosstab
table = pd.crosstab(filtered['Employment'], filtered['EdLevel'], normalize='index') * 100

# reorder columns to that order
table = table[ed_order]

# 4. plot
table.T.plot(kind='bar', stacked=True, figsize=(14,7), colormap='tab20')

plt.title("Top 5 Employment vs EdLevel (Sorted)")
plt.ylabel("Percentage")
plt.xlabel("Employment")
plt.xticks(rotation=45, ha='right')
plt.legend(title='EdLevel', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
