#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
df['JobSat'].median()
```

```python
critical_cols = ['Employment', 'JobSat', 'RemoteWork']
import pandas as pd
import numpy as np

print("Missing before:\n{}".format(df[critical_cols].isnull().sum()))
print("\n% Missing:\n{}".format(df[critical_cols].isnull().mean() * 100))

# 2. Strategy

# Employment - Very critical (2% missing usually) -> DROP rows, can't guess employment
# If someone didn't say employment status, whole row is not useful
df = df.dropna(subset=['Employment'])

# JobSat - Moderate importance (20-30% missing) -> Fill with mode, because satisfaction is categorical
df['JobSat'] = df['JobSat'].fillna(df['JobSat'].median())
# or fill with 'Unknown' if you want to keep missing as category
# df['JobSat'] = df['JobSat'].fillna('Unknown')

# RemoteWork - Important but not critical -> Fill with mode
mode_remote = df['RemoteWork'].mode()[0]
df['RemoteWork'] = df['RemoteWork'].fillna("unknown")

# 3. Verify
print("\nMissing after:\n{}".format(df[critical_cols].isnull().sum()))

```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
df[["YearsCodePro"]]=df[["YearsCodePro"]].replace({"More than 50 years":"51","Less than 1 year":"0"})
df["YearsCodePro"] = pd.to_numeric(df["YearsCodePro"], errors='coerce')
df["YearsCodePro"] = df["YearsCodePro"].fillna(df["YearsCodePro"].median())

df["JobSat"] = pd.to_numeric(df["JobSat"], errors='coerce')

exp_ranges=["0-5","5-10","10-20",">20"]
no_years=[
    (df["YearsCodePro"]<5),
    (df["YearsCodePro"]>=5) & (df["YearsCodePro"]<10),
    (df["YearsCodePro"]>=10) &(df["YearsCodePro"]<20),
    (df["YearsCodePro"]>=20)
]

df["ExperienceRanges"]=np.select(no_years,exp_ranges,default="other")

df_group=df.groupby("ExperienceRanges")[["JobSat"]].median().reset_index()
df_group
```

```python
## Write your code here


df_group.plot(kind='bar', x='ExperienceRanges', y='JobSat', legend=True)

# 3. Customize and display the plot
plt.title('Relationship between years of professional coding experience (YearsCodePro) and job satisfaction (JobSat)')
plt.xlabel('Experience Ranges')
plt.ylabel('Job satistfaction')
plt.show()
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
df_temp.head()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Reset everything to avoid duplicate labels error
df_temp = df[['Country', 'LanguageHaveWorkedWith']].copy()
df_temp.columns = ['Country', 'LanguageHaveWorkedWith'] # force unique column names
df_temp = df_temp.dropna().reset_index(drop=True)

df_temp['LanguageHaveWorkedWith'] = df_temp['LanguageHaveWorkedWith'].str.split(';')
df_temp = df_temp.explode('LanguageHaveWorkedWith')
df_temp['LanguageHaveWorkedWith'] = df_temp['LanguageHaveWorkedWith'].str.strip()
df_temp = df_temp[df_temp['LanguageHaveWorkedWith'] != '']

# Filter countries
top_countries = ['United States', 'India', 'Germany', 'United Kingdom', 'Canada']
df_temp = df_temp[df_temp['Country'].isin(top_countries)]

# Top 10 languages
top_langs = df_temp['LanguageHaveWorkedWith'].value_counts().head(10).index.tolist()
df_temp = df_temp[df_temp['LanguageHaveWorkedWith'].isin(top_langs)]

# USE GROUPBY + UNSTACK INSTEAD OF CROSSTAB - NO REINDEX ERROR
crosstab = df_temp.groupby(['Country', 'LanguageHaveWorkedWith']).size().unstack(fill_value=0)

# Sort by total
crosstab['total'] = crosstab.sum(axis=1)
crosstab = crosstab.sort_values(by='total', ascending=False).drop(columns='total')

print(crosstab)

# Plot
crosstab.plot(kind='bar', figsize=(16,7))
plt.title('Top 10 Languages by Country')
plt.xticks(rotation=20, ha='right')
plt.tight_layout()
plt.show()

plt.figure(figsize=(12,5))
sns.heatmap(crosstab, annot=True, fmt='d', cmap='Blues')
plt.title('Popularity by Region')
plt.tight_layout()
plt.show()
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Clean - keep only the 2 columns
df_temp = df[['EdLevel', 'Employment']].copy().dropna()

# For people with multiple employments, it shows like "Employed, full-time;Independent"
# Let's take the main one - split if you want to count each separately
# Simple version: keep as is


# 2. Sort EdLevel in logical order
order = [
    'Primary/elementary school',
    'Secondary school (e.g. American high school, German Realschule or Gymnasium, etc.)',
    'Some college/university study without earning a degree',
    'Associate degree (A.A., A.S., etc.)',
    'Bachelor’s degree (B.A., B.S., B.Eng., etc.)',
    'Master’s degree (M.A., M.S., M.Eng., MBA, etc.)',
    'Professional degree (JD, MD, Ph.D, Ed.D, etc.)',
    'Something else'
]
df_temp['EdLevel'] = pd.Categorical(df_temp['EdLevel'], categories=order, ordered=True)
df_temp = df_temp.sort_values('EdLevel')

top = df_temp['Employment'].value_counts().nlargest(5).index
df_temp['Emp_clean'] = df_temp['Employment'].apply(lambda x: x if x in top else 'Other')

# 3. CROSS-TAB - raw counts
# Crosstab
crosstab_pct = pd.crosstab(df_temp['EdLevel'], df_temp['Emp_clean'], normalize='index') * 100


```

```python
crosstab_pct
```

```python
# Heatmap - now only 6 cols
plt.figure(figsize=(12,5))
sns.heatmap(crosstab_pct.T, annot=True, cmap='Blues')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```

```python

# chart
crosstab_pct.plot(kind='bar', stacked=True, figsize=(10,5))
plt.legend(bbox_to_anchor=(1.02,1))
plt.tight_layout()
plt.show()

# What you'll see:

# Student % goes down as education goes up.
# Full-time % goes up as education goes up. Highest at Bachelor / Master.
# Not employed / Other is highest at low education (Primary / Secondary).
# So yes - higher education = more likely Full-time, lower education = more likely Student / Not employed.
```

```python
#prove correlation by calulating p value
#KNOWLEDGE about P value
# p < 0.001  = Extremely significant *** (your case)
# p < 0.01   = Very significant **
# p < 0.05   = Significant *  -> we say correlation exists
# p > 0.05   = Not significant -> no correlation, just randomfrom scipy.stats import chi2_contingency
chi2, p, _, _ = chi2_contingency(pd.crosstab(df_temp['EdLevel'], df_temp['Emp_clean']))
print("P value is: {}".format(p)) 
print("p < 0.001 = significant correlation")
```
