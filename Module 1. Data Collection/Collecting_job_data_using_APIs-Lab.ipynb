```python
#your code goes here
import requests
import pandas as pd

api_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json"
data = requests.get(api_url).json()

locations = ["Los Angeles", "New York", "San Francisco", "Washington DC", "Seattle", "Austin", "Detroit"]

def get_number_of_jobs_L(location):
    count = 0
    for job in data:
        if location.lower() in job.get("Location", "").lower():
            count += 1
    return count

results = []
for city in locations:
    results.append({"Location": city, "Number of Jobs": get_number_of_jobs_L(city)})


df = pd.DataFrame(results, columns=["Location", "Number of Jobs"])
df.to_excel("job-postings.xlsx", index=False)



print(df)
print("\nSaved to jobs_by_location.xlsx")
```

```python
# your code goes here
# your code goes here
api_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json"
data = requests.get(api_url).json()

skills = ["C", "C#", "C++", "Java", "JavaScript", "Python", "Scala", "Oracle", "SQL Server", "MySQL Server", "PostgreSQL", "MongoDB"]
wb = Workbook()                        # create a workbook object
ws = wb.active     # use the active worksheet

ws.append(["Key Skills", "Number of Jobs"])

# 4. Count postings per skill and write directly to Excel
for skill in skills:
    count = 0
    for job in data:
        # Match skill in Key Skills field
        if skill.lower() in job.get("Key Skills", "").lower():
            count += 1
    ws.append([skill, count])
    print("{}: {}".format(skill, count))

# 5. Save output
excel_filename = "job-postings-skills.xlsx"
wb.save(excel_filename)

print(f"Excel file created: {excel_filename}")
```
